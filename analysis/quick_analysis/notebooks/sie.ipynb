```python
def create_sample(input_path, output_path, sample_fraction=0.01):
    print(f"Sampling {sample_fraction*100}% of the data...")
    
    chunk_list = []
    
    reader = pd.read_csv(input_path, chunksize=100000, low_memory=False)
    
    for chunk in reader:
        sample_chunk = chunk.sample(frac=sample_fraction, random_state=42)
        chunk_list.append(sample_chunk)
    
    sample_df = pd.concat(chunk_list)
    
    sample_df.to_csv(output_path, index=False)
    print(f"Sample saved! New file contains {len(sample_df):,} rows.")
    return sample_df
```
